Clase Persona
Crea una clase Persona con los atributos

privados:

nombre
edad

Actividades:

 Crear métodos getters y setters
 Evitar que la edad sea negativa


```python
class Persona:
    def __init__(self, nombre, edad):
        self.__nombre = nombre
        # Usamos el setter aquí también para validar la edad al crear la persona
        self.set_edad(edad)

    # Getter para obtener el nombre
    def get_nombre(self):
        return self.__nombre

    # Setter para cambiar el nombre
    def set_nombre(self, nuevo_nombre):
        self.__nombre = nuevo_nombre

    # Getter para obtener la edad
    def get_edad(self):
        return self.__edad

    # Setter para cambiar la edad con VALIDACIÓN
    def set_edad(self, nueva_edad):
        if nueva_edad >= 0:
            self.__edad = nueva_edad
        else:
            print("Error: La edad no puede ser negativa. Se asignará 0.")
            self.__edad = 0
```

Ejercicio 2: Cuenta Bancaria
Crea una clase CuentaBancaria con:
 saldo (privado)
Actividades:
 Método para consultar saldo
 Método para depositar dinero
 Método para retirar dinero (solo si hay saldo
suficiente)

```python
class CuentaBancaria:
    def __init__(self, saldo_inicial):
        if saldo_inicial >= 0:
            self.__saldo = saldo_inicial
        else:
            self.__saldo = 0

    # Método para consultar saldo (funciona como un Getter)
    def consultar_saldo(self):
        return self.__saldo

    # Método para depositar dinero
    def depositar(self, cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
        else:
            print("No puedes depositar cantidades negativas.")

    # Método para retirar dinero con validación
    def retirar(self, cantidad):
        if cantidad <= self.__saldo:
            self.__saldo -= cantidad
            print(f"Retiro exitoso. Retiraste: ${cantidad}")
        else:
            print("Error: Saldo insuficiente.")
```

Ejercicio 3: Producto

Crea una clase Producto con:
 nombre
 precio

Actividades:
 El precio no puede ser negativo
 Crear método para aplicar descuento

```python
class Producto:
    def __init__(self, nombre, precio):
        self.__nombre = nombre
        if precio >= 0:
            self.__precio = precio
        else:
            self.__precio = 0

    def get_precio(self):
        return self.__precio

    # Método para aplicar un descuento (ejemplo: 10% se pasa como 10)
    def aplicar_descuento(self, porcentaje):
        if 0 < porcentaje <= 100:
            descuento = self.__precio * (porcentaje / 100)
            self.__precio -= descuento
            print(f"Descuento aplicado. Nuevo precio: ${self.__precio}")
        else:
            print("Porcentaje de descuento no válido.")
```

Ejercicio 4: Estudiante

Crea una clase Estudiante con:
 nombre
 nota final
Actividades:
 La nota debe estar entre 0 y 5
 Método para mostrar si aprueba o reprueba

```python
class Estudiante:
    def __init__(self, nombre, nota_final):
        self.__nombre = nombre
        self.set_nota_final(nota_final)

    def set_nota_final(self, nota):
        if 0 <= nota <= 5:
            self.__nota_final = nota
        else:
            print("Error: La nota debe estar entre 0 y 5. Se asigna 0.")
            self.__nota_final = 0

    def mostrar_estado(self):
        if self.__nota_final >= 3.0:
            return f"El estudiante {self.__nombre} ¡APROBÓ! con {self.__nota_final}"
        else:
            return f"El estudiante {self.__nombre} REPROBÓ con {self.__nota_final}"
```

Ejercicio 5: Vehículo

Crea una clase Vehiculo con:
 velocidad
Actividades:
 No permitir velocidades negativas
 Método para acelerar y frenar

```python
class Vehiculo:
    def __init__(self):
        self.__velocidad = 0  # Empieza detenido

    def get_velocidad(self):
        return self.__velocidad

    def acelerar(self, incremento):
        if incremento > 0:
            self.__velocidad += incremento
        else:
            print("El incremento debe ser positivo.")

    def frenar(self, decremento):
        if decremento > 0:
            # Evitamos que al restar la velocidad baje de 0
            if self.__velocidad - decremento >= 0:
                self.__velocidad -= decremento
            else:
                self.__velocidad = 0
        else:
            print("El decremento debe ser positivo.")
```

Ejercicio 7: Rectángulo
Crea una clase Rectangulo con:
 base
 altura
Actividades:
 Validar que los valores sean positivos
 Crear métodos para calcular área y perímetro

```python
class Rectangulo:
    def __init__(self, base, altura):
        if base > 0 and altura > 0:
            self.__base = base
            self.__altura = altura
        else:
            print("Valores incorrectos. Se creará un rectángulo de 1x1.")
            self.__base = 1
            self.__altura = 1

    def calcular_area(self):
        return self.__base * self.__altura

    def calcular_perimetro(self):
        return 2 * (self.__base + self.__altura)
```

Ejercicio 8: Libro
Crea una clase Libro con:
 título
 número de páginas
Actividades:
 No permitir páginas negativas
 Método para mostrar información del libro

```python
class Libro:
    def __init__(self, titulo, num_paginas):
        self.__titulo = titulo
        if num_paginas >= 0:
            self.__num_paginas = num_paginas
        else:
            print("Error: Un libro no puede tener páginas negativas. Se asigna 0.")
            self.__num_paginas = 0

    def mostrar_informacion(self):
        print(f"Libro: '{self.__titulo}' | Páginas: {self.__num_paginas}")
```

Ejercicio 9: Temperatura
Crea una clase Temperatura con:
 valor en grados Celsius
Actividades:
 Método para convertir a Fahrenheit
 Validar que el valor no sea menor a -273.15

```python
class Temperatura:
    def __init__(self, celsius):
        if celsius >= -273.15:
            self.__celsius = celsius
        else:
            print("Error: Temperatura por debajo del cero absoluto. Se asigna -273.15")
            self.__celsius = -273.15

    def convertir_a_fahrenheit(self):
        # Fórmula: (C * 9/5) + 32
        return (self.__celsius * 9/5) + 32
```

Ejercicio 10: Sistema de
Inventario
Crea una clase Inventario con:
 cantidad de productos
Actividades:
 No permitir valores negativos
 Método para agregar productos
 Método para retirar productos (sin quedar negativo)

```python
class Inventario:
    def __init__(self, cantidad_inicial):
        if cantidad_inicial >= 0:
            self.__cantidad_productos = cantidad_inicial
        else:
            self.__cantidad_productos = 0

    def get_cantidad(self):
        return self.__cantidad_productos

    def agregar_productos(self, cantidad):
        if cantidad > 0:
            self.__cantidad_productos += cantidad
        else:
            print("Debes agregar una cantidad mayor a cero.")

    def retirar_productos(self, cantidad):
        if cantidad > 0:
            if self.__cantidad_productos - cantidad >= 0:
                self.__cantidad_productos -= cantidad
                print(f"Se retiraron {cantidad} productos.")
            else:
                print("Error: No puedes retirar más productos de los que hay en existencia.")
        else:
            print("Cantidad a retirar no válida.")
```
