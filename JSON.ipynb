````python
import json

# 1. Definimos la cadena de texto con formato JSON
data_json = '''
[
  {
    "nombre": "Cristiano",
    "apellido": "Ronaldo",
    "edad": 36,
    "equipos": ["Sporting Club", "Manchester United", "Real Madrid", "Juventus"]
  },
  {
    "nombre": "Lionel",
    "apellido": "Messi",
    "edad": 34,
    "equipos": ["Newell's Old Boys", "FC Barcelona", "PSG"]
  },
  {
    "nombre": "Zlatan",
    "apellido": "Ibrahimovic",
    "edad": 40,
    "equipos": ["Malmö FF", "Ajax", "Juventus", "Inter", "Milan", "PSG"]
  }
]
'''

# 2. Convertimos el JSON a una lista de diccionarios de Python
jugadores = json.loads(data_json)

# 3. Ejemplo de cómo acceder a los datos
for jugador in jugadores:
    print(f"Jugador: {jugador['nombre']} {jugador['apellido']} - Edad: {jugador['edad']}")
    print(f"Último equipo: {jugador['equipos'][-1]}")
    print("-" * 30)
````

````python
import json  # Importamos la librería para poder transformar la lista a formato JSON

# 1. Definimos el conjunto de arreglos (Matriz de datos)
# Estructura: [Nombre, Apellido, Edad, [Equipos]]
datos_jugadores = [
    ["Cristiano", "Ronaldo", 36, ["Sporting Club", "Manchester United", "Real Madrid", "Juventus"]],
    ["Lionel", "Messi", 34, ["Newell's Old Boys", "FC Barcelona", "PSG"]],
    ["Zlatan", "Ibrahimovic", 40, ["Malmö FF", "Ajax", "Juventus", "Inter", "Milan", "PSG"]]
]

# 2. Convertir nuestra lista de Python a una "String" con formato JSON
# Esto es lo que enviarías a una página web o a una base de datos
json_formateado = json.dumps(datos_jugadores, indent=2)

# 3. Mostrar el resultado
print(" DATOS EN FORMATO JSON (STRING)")
print(json_formateado)

print("ACCESO MANUAL A LOS DATOS")
# Para obtener el nombre de Messi:
# Buscamos el segundo arreglo (índice 1) y su primer elemento (índice 0)
nombre_messi = datos_jugadores[1][0]
apellido_messi = datos_jugadores[1][1]

# Para obtener el último equipo de Zlatan (Tercer arreglo, índice 2 / Lista de equipos, índice 3)
ultimo_equipo_zlatan = datos_jugadores[2][3][-1]

print(f"Jugador: {nombre_messi} {apellido_messi}")
print(f"Último equipo de Zlatan: {ultimo_equipo_zlatan}")
````

````python
import json

datos = {"nombre": "Cursor", "version": 2026, "activo": True}

# 1. ESCRIBIR (DUMP)
with open("config.json", "w") as archivo:
    # indent=4 hace que el archivo sea legible para humanos
    json.dump(datos, archivo, indent=4)

# 2. LEER (LOAD)
with open("config.json", "r") as archivo:
    datos_leidos = json.load(archivo)
    print(datos_leidos["nombre"]) # Resultado: Cursor
````
