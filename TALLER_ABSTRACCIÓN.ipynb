#7.  	Uso de atributos
Crea una clase que represente un estudiante.

La clase debe tener: un atributo llamado nombre, un atributo llamado edad
Luego crea un método llamado presentarse()
El método debe mostrar:
"Hola, soy un estudiante y tengo 18 años"


```python
class Estudiante:
    # El constructor recibe los datos y los guarda en el objeto
    def __init__(self, nombre, edad):
        self.nombre = nombre  # Atributo nombre
        self.edad = edad      # Atributo edad

    def presentarse(self):
        # Usamos f-string (la 'f' antes de las comillas) para meter las variables en el texto
        print(f"Hola, soy un estudiante y tengo {self.edad} años")

# Prueba del código:
# Creamos al estudiante pasándole el nombre y la edad (18 como pide el ejercicio)
nuevo_estudiante = Estudiante("Carlos", 18)
nuevo_estudiante.presentarse()
```
